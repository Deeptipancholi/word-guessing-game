```python
import random

class WordGuessingGame:

    def __init__(self):
        self.difficulty = None
        self.word = None
        self.chances = 5
        self.score = 0

        self.words = {
            "Easy": {
                "cat": [
                    "It is an animal.",
                    "It has four legs.",
                    "It always lands on her feet."
                ],
                "bat": [
                    "It can fly.",
                    "It is active at night.",
                    "It hangs upside down."
                ],
                "apple": [
                    "It is a fruit.",
                    "It can be red or green.",
                    "It starts with A."
                ],
                "ball": [
                    "It is round.",
                    "You can play with it.",
                    "It is used in many sports."
                ]
            },

            "Medium": {
                "kitchen": [
                    "It is a part of a house.",
                    "Food is prepared here.",
                    "It contains a stove."
                ],
                "cutlery": [
                    "I am the metal team found at your dinner table.",
                    "Without us, eating a meal would be difficult.",
                    "Find me where plates and food are served."
                ],
                "rainbow": [
                    "You can see it after rain.",
                    "It has many colors.",
                    "It appears in the sky."
                ],
                "spices": [
                    "They are used in cooking.",
                    "They add flavor to food.",
                    "Chilli is one example."
                ]
            },

            "Hard": {
                "architecture": [
                    "It is related to buildings.",
                    "It involves designing structures.",
                    "Architects work in this field."
                ],
                "algorithm": [
                    "It is related to programming.",
                    "It is a step-by-step procedure.",
                    "It is used to solve problems."
                ],
                "psychology": [
                    "It is related to the human mind.",
                    "It studies behavior.",
                    "It is a field of science."
                ],
                "astronomy": [
                    "It is related to space.",
                    "It studies stars and planets.",
                    "Scientists study the universe."
                ]
            }
        }

    def choose_difficulty(self):

        print("Choose your Difficulty Level:")
        print("1. Easy")
        print("2. Medium")
        print("3. Hard")

        choice = input("Enter your difficulty level: ").strip().title()

        if choice == "Easy":
            self.difficulty = "Easy"

        elif choice == "Medium":
            self.difficulty = "Medium"

        elif choice == "Hard":
            self.difficulty = "Hard"

        else:
            print("Invalid choice")
            return False

        return True

    def select_word(self):

        self.word = random.choice(
            list(self.words[self.difficulty].keys())
        )

    def calculate_score(self, chance):

        if self.difficulty == "Easy":
            points = [30, 24, 18, 12, 6]

        elif self.difficulty == "Medium":
            points = [50, 40, 30, 20, 10]

        else:
            points = [70, 56, 42, 28, 14]

        self.score = points[chance]

    def guess_word(self):

        for chance in range(self.chances):

            hint = self.words[self.difficulty][self.word][chance % 3]

            print("\nHint:", hint)

            guess = input("Guess the word: ").strip().lower()

            if guess == self.word.lower():

                print("Correct! 🎉")

                self.calculate_score(chance)

                print("You earned", self.score, "points!")

                break

            else:

                print("Wrong guess ❌")
                print("Chance left:", self.chances - chance - 1)

        else:

            print("You lost! 😢")
            print("The correct word was:", self.word)


# ---------------- GAME START ----------------

while True:

    game = WordGuessingGame()

    if game.choose_difficulty():

        game.select_word()

        game.guess_word()

        print("\nSelected difficulty:", game.difficulty)
        print("Your final score:", game.score)

        play_again = input("\nDo you want to play again? (yes/no): ").strip().lower()

        if play_again != "yes":
            print("\nThanks for playing! 🎮")
            break
```
